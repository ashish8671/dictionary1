**List Comprehension**

```python
[body_of_for for_loop]
[body_of_if for_loop if condition]
```

```python
l=[]
for i in range(1,5):
    l.append(1**2)
```

```python
[i**2 for i in range(1,10,2)]
```

```python
print(round(1.5678,0))
print(round(1.5678,1))
print(round(1.5678,2))
```

```python
# round all elements in the list

l = [1.3, 7.8888, 5.4, 9.50000]
list(map(round, l))
```

```python
# def square(x): return x*x
list(map(lambda x: x/6, [1,2,3]))
```

```python
# remove all the odd numbers from list

list(map(lambda x: x%2==0, [1,2,3]))
```

```python
list(filter(lambda x: x%2==0, [1,2,3,4,5,6]))
```
