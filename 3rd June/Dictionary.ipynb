```python
def replace_mood (sentence):
    words=sentence.split()
    result=""
    for i in words:   # scan the everu word
        if i in emojis:  #is the word present
            result+=emojis[i]  # get the value
        else: result+=i
    return result

emojis = {
    "happy":"😊",
    "sad":"😒",
    "love":"😍",
    "frustrated":"😫"
}
sentence="i am happy"
print(replace_mood (sentence))
```

```python
import matplotlib.pyplot as plt
from collections import Counter

s = "Ashish"

freq = {}

for ch in s.lower():   # convert in lower case
    if ch.isalpha():   # whether the letter is alphabet
        freq[ch] = freq.get(ch, 0) + 1   # frequency cecking statement

plt.bar(freq.keys(), freq.values())    # create a bar chart 
plt.xlabel("Letters")   # label X axis as a Letter
plt.ylabel("Frequency")  # label Y axis as a Frequency
plt.title("Letter Frequency")  # label title as Letter Frequency
plt.show()   # to show the graph
```
